# 查看dataloder形式

```python
import torch
import numpy as np
from pprint import pprint

# 假设你刚才定义的类在当前脚本中
from prediff.datasets.sevir.sevir_torch_wrap import NPYSEVIRDataLoader

npy_dir = "/home/user01/25fall_aiclass/lesson_resource/data/prediff/datasets/sevirlr/data_npy"

loader = NPYSEVIRDataLoader(
    npy_dir=npy_dir,
    seq_len=13,
    stride=1,
    batch_size=2,
    layout="NHWT",
    preprocess=False,   # 建议先关掉预处理，看原始数据
    verbose=True
)

# 看看基本信息
print(f"\n=== Loader Info ===")
print(f"总事件数: {loader.total_num_event}")
print(f"每事件可生成序列数: {loader.num_seq_per_event}")
print(f"总子序列数: {loader.total_num_seq}")
print(f"__len__(): {len(loader)}")

# 取一个 batch
batch_idx = 6
batch = loader._idx_sample(batch_idx)

print(f"\n=== _idx_sample({batch_idx}) 输出 ===")
for k, v in batch.items():
    print(f"{k}: type={type(v)}, shape={v.shape}, dtype={v.dtype}, min={v.min().item():.3f}, max={v.max().item():.3f}")

# 打印该 batch 内每个样本对应的文件路径
print(f"\n=== 本批样本对应的文件 ===")
for i in range(loader.batch_size):
    event_idx = (batch_idx * loader.batch_size) // loader.num_seq_per_event + i // loader.num_seq_per_event
    fname = loader._samples.iloc[event_idx]['vil_filename']
    print(f"样本 {i}: {fname}")
```
